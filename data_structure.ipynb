# Series(1D Table)

```python
import pandas as pd

data = ['shubam','devansh','prarambh','sanjay',
        'rajeev','rajesh','gaurav','mekail','james',
        'nikovich']

s = pd.Series(data)

print(s)
```

```python
print(pd.__version__)
```

# Creating Series with custom index

```python
data = [10,20,30,40,50,60]
s = pd.Series(data)
print(s[0])
s = pd.Series(data,index=["a","b","c","d","e","f"])
print(s)
```

# Series from dictionary

```python
data = {
    "Aman": 85,
    "Sara": 90,
    "Rahul": 78,
    "Tammana": 85,
    "Riya": 90,}
s = pd.Series(data)
print(s)
```

# Series Operations

```python
s = pd.Series([1,2,3,4,5])
print(s+10)
```

# DataFrame(2D Table)

```python
data = {
  "calories": [420, 380, 390, 350, 410],
  "duration": [50, 40, 45, 50,40]
}

#load data into a DataFrame object:
df = pd.DataFrame(data)

print(df) 
```

```python
# refer to the row index:
print(df.loc[0])
```

```python
# use a list of indexes:
print(df.loc[[0,1]])
```

```python
df = pd.DataFrame(data, index=['day1','day2','day3','day4','day5'])
print(df)
```

```python
print(df.loc['day2'])
```
